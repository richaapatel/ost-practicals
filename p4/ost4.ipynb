```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("BostonHousing.csv")
```

```python
print("Basic Information:")
print(df.info())
print(df.describe())
print("\nFirst 5 Rows:")
display(df.head())
```

```python
avg_medv = df.groupby("rad")["medv"].mean()

plt.figure(figsize=(8, 5))
plt.bar(avg_medv.index.astype(str), avg_medv.values)
plt.xlabel("RAD")
plt.ylabel("Average MEDV")
plt.title("Average House Price by RAD")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["rm"], df["medv"])
plt.xlabel("Average Number of Rooms (RM)")
plt.ylabel("House Price (MEDV)")
plt.title("RM vs MEDV")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["medv"], bins=20, edgecolor="black")
plt.xlabel("House Price (MEDV)")
plt.ylabel("Frequency")
plt.title("Distribution of House Prices")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df.index[:50], df["medv"][:50])
plt.xlabel("Record Number")
plt.ylabel("House Price (MEDV)")
plt.title("House Prices for First 50 Records")
plt.show()
```

```python
rad_counts = df["rad"].value_counts().sort_index()

plt.figure(figsize=(8, 8))
plt.pie(
    rad_counts.values,
    labels=rad_counts.index,
    autopct="%1.1f%%"
)
plt.title("Distribution of Houses by RAD")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.boxplot(df["medv"])
plt.ylabel("House Price (MEDV)")
plt.title("Box Plot of House Prices")
plt.show()
```
